```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(7)
```

# Data Cell

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

sales = [12000, 15000, 13500, 17000, 16000, 19000,
         18500, 21000, 20000, 22500, 24000, 26000]

subjects = ['Math', 'Science', 'English', 'History', 'Art']
scores   = [78, 85, 72, 90, 65]

study_hours = np.random.uniform(1, 10, 100)
exam_scores = study_hours * 8 + np.random.normal(0, 5, 100)

product_sales = np.random.normal(loc=50, scale=10, size=200)
```

**Task 1 — Line Plot: Monthly Sales Trend**

```python
plt.figure()

plt.plot(months, sales, marker='o')

plt.title("Monthly Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

*My Explanation*

👉 First I used months and sales data

👉 Then I plotted a line graph

👉 Markers help to see each data point

👉 Title and labels make graph easy to understand

**Task 2 — Bar Plot: Subject Score Comparison**

```python
plt.figure()

plt.bar(subjects, scores)

plt.title("Subject Score Comparison")
plt.xlabel("Subjects")
plt.ylabel("Scores")

plt.show()
```

*My Explanation*

👉 I used bar plot to compare subjects

👉 Each bar shows score of one subject

👉 Easy to compare visually

**Task 3 — Scatter Plot: Study Hours vs Exam Score**

```python
sns.scatterplot(x=study_hours, y=exam_scores)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.show()
```

*My Explanation*

👉 Scatter plot shows relationship

👉 More study hours → higher score trend

👉 Each point = one student

**Task 4 — Box Plot: Product Sales Distribution**

```python
sns.boxplot(x=product_sales)

plt.title("Product Sales Distribution")
plt.xlabel("Sales")

plt.show()
```

*My Explanation*

👉 Box plot shows:

median

spread

outliers


👉 Helps understand distribution
